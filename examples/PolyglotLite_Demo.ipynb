## Using Pretrained Models

The easiest way to get started. Uses HuggingFace models as backend.

```python
from polyglotlite import PolyglotLiteHF

model = PolyglotLiteHF("polyglot-135m")
print(model)
```

```python
# Generate text
output = model.generate(
    "The future of artificial intelligence is",
    max_length=50,
    temperature=0.7
)
print(output)
```

```python
# Try different prompts
prompts = [
    "Machine learning can help us",
    "The best way to learn programming is",
    "In the year 2050,",
]

for prompt in prompts:
    output = model.generate(prompt, max_length=40)
    print(f"Prompt: {prompt}")
    print(f"Output: {output}\n")
```

## Language Detection

```python
from polyglotlite import detect_language, get_supported_languages

languages = get_supported_languages()
print(f"Supported languages: {len(languages)}")

texts = [
    "Hello, how are you?",
    "Bonjour, comment allez-vous?",
    "Hola, como estas?",
    "Guten Tag, wie geht es Ihnen?",
]

for text in texts:
    lang = detect_language(text)
    print(f"{text} -> {lang} ({languages.get(lang, 'Unknown')})")
```

## Training Your Own Model

You can also train from scratch on your own data.

```python
from polyglotlite import PolyglotLite, Trainer

# Initialize untrained model
model = PolyglotLite(model_name="polyglot-135m")
print(model)

# Sample training data
train_texts = [
    "PolyglotLite is a multilingual language model.",
    "It supports over 50 languages.",
    "The model runs on consumer hardware.",
    "Fine-tuning is straightforward.",
] * 10

# Create trainer (not actually running here)
trainer = Trainer(
    model=model,
    train_data=train_texts,
    batch_size=2,
    max_steps=100,
    learning_rate=1e-4,
)

print("Trainer configured. Call trainer.train() to start.")
```
