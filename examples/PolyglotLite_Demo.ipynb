# 🌍 PolyglotLite: Efficient Multilingual Language Models

This notebook demonstrates how to use PolyglotLite for multilingual text generation.

**Features:**
- Lightweight models (135M-500M parameters)
- Support for 100+ languages
- Runs on consumer hardware
- Easy fine-tuning with QLoRA

## 2. Quick Start

```python
from polyglotlite import PolyglotLite

# Load the model
model = PolyglotLite.from_pretrained("polyglot-135m")
print(model)
```

```python
# Generate text
output = model.generate(
    "The future of AI is",
    max_length=50,
    temperature=0.7
)
print(output)
```

## 3. Language Detection

```python
from polyglotlite import detect_language, get_supported_languages

# Check supported languages
languages = get_supported_languages()
print(f"Supported languages: {len(languages)}")

# Detect language
texts = [
    "Hello, how are you?",
    "Bonjour, comment allez-vous?",
    "你好，你好吗？",
    "مرحبا كيف حالك",
]

for text in texts:
    lang = detect_language(text)
    print(f"{text} -> {lang} ({languages.get(lang, 'Unknown')})")
```

## 4. Multilingual Generation

```python
# Generate in different languages
prompts = {
    "English": "Once upon a time",
    "French": "Il était une fois",
    "Spanish": "Érase una vez",
    "German": "Es war einmal",
}

for lang, prompt in prompts.items():
    output = model.generate(prompt, max_length=30)
    print(f"\n{lang}:")
    print(f"  {output}")
```

## 5. Fine-tuning Example

```python
from polyglotlite import Trainer, TrainingConfig

# Sample training data
train_texts = [
    "PolyglotLite is a multilingual language model.",
    "It supports over 100 languages.",
    "Fine-tuning is easy with QLoRA.",
    "The model runs on consumer hardware.",
] * 10  # Repeat for demo

# Create trainer
trainer = Trainer(
    model=model,
    train_data=train_texts,
    batch_size=2,
    max_steps=100,
    learning_rate=1e-4,
)

# Note: Run trainer.train() to actually train
# trainer.train()
print("Trainer configured! Call trainer.train() to start training.")
```

## 6. Save and Export

```python
# Save model
model.save_pretrained("./my-polyglot-model")

# Export to ONNX for deployment
# model.export_onnx("model.onnx")

print("Model saved!")
```
